## 第一章 线性代数中的线性方程组


许多重要的管理决策是在线性规划模型的基础上做出的，这些模型包含了几百个变量，  
例如，航运业使用线性规划调度航班，监视飞机的飞行位置，或计划维修和机场运行；


#### 1.1 线性方程组


线性方程组就是形如 $a_1 x_1 + a_2 x_2 + \dots + a_n x_n = b$ 的方程；  
而 $4x_1 - 5x_2 = x_1x_2$ 和 $x_2 = 2\sqrt{x_1} - 6$ 都不是线性方程；

线性方程组所有的解的集合称为线性方程组的解集，若两个线性方程组有相同的解集，则两个线性方程组是等价的；  
求包含两个变量的线性方程组成的方程组的解，等价于求两条直线的交点；  
如 $x_1 - 2x_2 = -1$ 和 $-x_1 + 3x_2 = 3$


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['SimHei']  # 或 ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False    # 解决负号显示问题

# 方程：
# 1) x1 - 2x2 = -1  →  x1 = 2x2 - 1
# 2) -x1 + 3x2 = 3  →  x1 = 3x2 - 3

# 定义 x2 的取值范围
x2 = np.linspace(-5, 5, 400)

# 计算对应的 x1
x1_line1 = 2 * x2 - 1
x1_line2 = 3 * x2 - 3

# 求交点（解方程组）
# x1 - 2x2 = -1
# -x1 + 3x2 = 3
A = np.array([[1, -2],
              [-1, 3]])
b = np.array([-1, 3])

x1_sol, x2_sol = np.linalg.solve(A, b)

# 绘图
plt.figure(figsize=(6, 6))
plt.plot(x2, x1_line1, label=r"$x_1 - 2x_2 = -1$")
plt.plot(x2, x1_line2, label=r"$-x_1 + 3x_2 = 3$")

# 标注交点
plt.scatter(x2_sol, x1_sol, color='red', zorder=5)
plt.text(x2_sol + 0.1, x1_sol + 0.1,
         f"({x2_sol:.2f}, {x1_sol:.2f})",
         color='red')

# 坐标轴设置
plt.xlabel("$x_2$")
plt.ylabel("$x_1$")
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.grid(True)
plt.legend()
plt.title("两条直线的交点")

plt.show()
```

我们称一个线性方程组是相容的，若它有一个解或无穷多个解；  
称它是不相容的，若它无解；
