## 第一章 理解大语言模型

- 大语言模型，作为深度神经网络模型的代表，为NLP领域带来了革命性的变化。       
- 我们谈到LLM的“理解”能力时，实际上是它们能够处理看似连贯且符合语境的文本。       
- 深度学习使得LLM能够利用海量的文本数据来训练，捕获更深层次的上下文联系。     
- 一方面得益于Transformer架构，另一方面得益于海量数据。     
- 流行的LLM用Transformer，但并非LLM就等于Transformer。

### 1.1 什么是大语言模型

- 大语言模型本质上是神经网络，属于深度神经网络（deep neural network）。     
- LLM一般都拥有数百亿及以上的参数parameter，越高越有利于“下一个单词预测”。            
- Transformer架构使得其更擅长应对人类语言的细微差别。       
- 深度学习利用3层及以上的神经网络来建模数据中的复杂模式和抽象特征。     
- 以垃圾邮件识别为例，传统机器学习由人定义特征，而深度学习则是人标记一部分垃圾邮件，然后由机器自己学习。      

### 1.2 大语言模型的应用

- 几乎所有的需要解析和生成文本的任务的自动化处理任务中都具有重要价值。      

### 1.3 构建和使用大语言模型的各个阶段

- 针对特定领域量身打造的LLM往往在性能上由于通用LLM。        
- LLM的构建通常包括pre-trainning和fine-tuning。     
- 预训练的预是指初始阶段，此时在大规模，多样化的数据上训练，形成全面的语言理解能力。        
- 微调阶段会在规模较小的数据上训练特定任务的能力。      
- 预训练后的LLM通常叫做基础模型，即foundation model。       
- 微调阶段用带有标注的数据集进一步训练，目前流行两种方法，指令微调和分类任务微调。      
- 指令微调instruction fine-tuning，标注数据集由“指令-答案”对组成。      
- 分类任务微调classification fine-tuning，标注数据集由文本及其类别标签组成。

### 1.4 Transformer架构

- 那幅著名的截图展示的是一段句子的翻译过程的最后阶段。        
- Transformer架构由2个子模块构成，编码器和解码器；一个关键组件自注意力机制。
- 编码器encoder模块负责处理输入文本，将其编码为一系列数值表示或向量。
- 解码器decoder模块负责接收这些编码向量，并据此生成输出文本。
- 编码器解码器都由多层组成，这些层通过自注意力机制连接。
- 自注意力机制self-attention mechainism，允许模型衡量序列中不同单词和词元之间的相对重要性。
- Transformer后续变体主要有BERT和各种GPT。
- BERT即Bidirectional Encoder Representations from Transformer，基于Transformer的双向编码器表示。
- BERT基于原始Transformer的编码器模块构建，训练方法与GPT不同，专注于掩码预测masked word prediction。
- BERT在情感预测，文档分类任务中具有优势。
- 各种GPT侧重于原始Transformer架构中的解码器部分，主要用于处理生成文本的任务，机器翻译，文本摘要，小说写作，代码编写等。
- 零样本学习zero-shot learning指在没有任何提示时泛化到从未见过的任务。
- 少样本学习few-shot learning 指从用户提供的少量样本中学习。

### 1.5 利用大型数据集

- GPT-3是第一代ChatGPT的基础模型。
- 词元token是模型读取文本的基本单位。训练LLM数据集中的词元数量大致等于单词数加上标点符号数。
- 训练GPT-3的数据集大约有4990亿的词元，约570G，但最终训练时候只用了3000亿；
- 训练GPT-3的云计算成本高达460万美元。

### 1.6 深入剖析GPT架构

- 下一个单词预测任务采用的是自监督学习self-supervised learning模式。    
- 与原始Transformer架构比，GPT的通用架构更简洁，只包含解码器部分，不包含编码器。
- GPT这样解码器模型是逐词预测生成文本，被认为是一种自回归模型autoregressive model。
- GPT-3共有96层Transformer和1750亿个参数，而原始的Transformer模型只将编码器和解码器模块重复了6次。
- 涌现emergence，模型能够完成未经明确训练任务的能力。

### 1.7 构建大语言模型

- 3个阶段，1.实现模型构架和准备数据，2.预训练大语言模型以获得基础模型，3.微调基础模型以得到个人助手或文本分类器。
- 第一阶段主要任务包括了实现出大语言模型的核心组件——注意力机制。
